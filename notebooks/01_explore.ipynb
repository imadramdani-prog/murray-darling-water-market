## Water allocation prices, 2001–2021

*Years are water years ending in June (2008 = July 2007 to June 2008).
Price is the simple average of the 9 southern regions, in real 2019-20 dollars.*

Prices move in clear cycles that follow water scarcity:

- **2008: record high (~$690/ML).** The peak of the Millennium Drought.
  Allocations were cut sharply, so irrigators competed for very little water.
- **2011–2012: record low (~$25–40/ML).** After the 2010–11 floods, dams refilled
  and allocations recovered. Prices fell about 28 times from the 2008 peak.
- **2020: second peak (~$525/ML).** The 2017–2019 drought, with 2019 the driest
  year on record in the Basin.

**Open question:** in 2020, high-reliability allocations were much higher than
in 2008 (87% vs 56%), yet prices were almost as high. The relationship between
water supply and price may have changed. The rest of this analysis tests it.

When high-reliability allocations were almost full (95% or more), the average price was about $122/ML before 2016 and $229/ML after: nearly double. 
This needs to be confirmed by a model that controls for other factors

## Model data

```python
cols = ["Year", "Region", "P", "R", "a_High", "a_Low",
        "C_High", "C_Low", "E_High", "E_Low", "D_High", "D_Low"]
model_df = df_south[cols].copy()
model_df.shape
```

```python
(model_df["C_High"] < 0).sum()
```

```python
model_df["C_High"] = model_df["C_High"].clip(lower=0)
```

```python
import numpy as np

model_df["log_P"] = np.log(model_df["P"])
model_df["after_2016"] = (model_df["Year"] >= 2017).astype(int)
model_df["eff_High"] = model_df["a_High"] * (1 - model_df["D_High"])
model_df["eff_Low"] = model_df["a_Low"] * (1 - model_df["D_Low"])
model_df["carry_share"] = (model_df["C_High"] + model_df["C_Low"]) / (model_df["E_High"] + model_df["E_Low"])

model_df.shape
```
