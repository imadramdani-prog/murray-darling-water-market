## Water allocation prices, 2001–2021

*Years are water years ending in June (2008 = July 2007 to June 2008).
Price is the simple average of the 9 southern regions, in real 2019-20 dollars.*

Prices move in clear cycles that follow water scarcity:

- **2008: record high (~$690/ML).** The peak of the Millennium Drought.
  Allocations were cut sharply, so irrigators competed for very little water.
- **2011–2012: record low (~$25–40/ML).** After the 2010–11 floods, dams refilled
  and allocations recovered. Prices fell about 28 times from the 2008 peak.
- **2020: second peak (~$525/ML).** The 2017–2019 drought, with 2019 the driest
  year on record in the Basin.

**Open question:** in 2020, high-reliability allocations were much higher than
in 2008 (87% vs 56%), yet prices were almost as high. The relationship between
water supply and price may have changed. The rest of this analysis tests it.

```python

df_south.plot.scatter(x="a_High", y="P", title="Water allocation price / high-reliability allocation, southern MDB, 2001-2021", xlabel="High-reliability allocation", ylabel="Price ($/ML, 2019-20 dollars)", figsize=(10, 5))
```

```python
before = df_south[df_south["Year"] <= 2016]
```

```python
after = df_south[df_south["Year"] > 2016]
```

```python
ax = before.plot.scatter(
    x="a_High", y="P", color="tab:blue", label="2001-2016",
    title="Price vs high-reliability allocation, before and after 2016",
    xlabel="High-reliability allocation (share of entitlement)",
    ylabel="Price ($/ML, 2019-20 dollars)",
    figsize=(10, 5),
)
after.plot.scatter(x="a_High", y="P", color="tab:orange", label="2017-2021", ax=ax)
ax.legend();
```

```python
from matplotlib.ticker import PercentFormatter

ax = before.plot.scatter(x="a_High", y="P", color="tab:blue", label="2001-2016", figsize=(10, 5))
after.plot.scatter(x="a_High", y="P", color="tab:orange", label="2017-2021", ax=ax)

ax.set_title("Price vs high-reliability allocation, southern MDB, before and after 2016")
ax.set_xlabel("High-reliability allocation (% of entitlement)")
ax.set_ylabel("Price ($/ML, 2019-20 dollars)")
ax.xaxis.set_major_formatter(PercentFormatter(1.0))
ax.legend();
```
